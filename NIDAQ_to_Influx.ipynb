# NI-DAQ to InfluxDB

The logger has moved to the `nidaq_to_influx` Python package. See [README.md](README.md) for installation and configuration.

After installation, set `INFLUX_TOKEN` in your environment and run `nidaq-to-influx` from any terminal directory. Use `--print-default-config` to create an editable TOML configuration and `--config /absolute/path/to/config.toml` to load it. Press Ctrl+C to stop and drain pending writes.

Default settings are in [src/nidaq_to_influx/default.toml](src/nidaq_to_influx/default.toml). The token is never stored in configuration.
